```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
import sklearn.metrics as metrics
import statsmodels.api as sm
import sklearn.model_selection as model_selection
```

# **Build Simple Regreassion  Linear model**

---



```python
study_hours=[1,2,3,4,5,6,7,8]
Exam_score=[5,6,4,8,7,9,7,5]
```

```python
X_constant = sm.add_constant(study_hours)
full_model = sm.OLS(Exam_score, X_constant).fit()
print(full_model.summary())
```

```python
# Convert lists to pandas Series to use the .drop method
x_train = [1,2,3,4,5,6]
y_train = [5,6,4,8,7,9]
```

```python
# Aligning both training lists to have the same length (e.g., first 6 elements)
x_train_aligned = x_train[:min(len(x_train), len(y_train))]
y_train_aligned = y_train[:min(len(x_train), len(y_train))]

# Fit the OLS model using the aligned data
X_constant = sm.add_constant(x_train_aligned)
trained_model = sm.OLS(y_train_aligned, X_constant).fit()
print(trained_model.summary())
```

## **In the next step work we need to make a Partial Comparison and Full Comparison between using MSE and RMSE**


```python
# Define test datasets
x_test = [7, 8]
y_test = [7, 5]

# Add the constant term to the test features matching the model format
X_test_constant = sm.add_constant(x_test, has_constant='add')

# Make predictions using the trained model
y_pred = trained_model.predict(X_test_constant)

# Create a comparison table
comparison_df = pd.DataFrame({
    'Study Hours (x)': x_test,
    'Actual Score (y)': y_test,
    'Predicted Score': y_pred,
    'Error (Residual)': np.array(y_test) - y_pred
})

print("Partial Comparison between Test Values and Model Predictions:")
print(comparison_df.to_string(index=False))
```

```python
import numpy as np
import sklearn.metrics as metrics

# Calculate Partial MSE and RMSE
partial_mse = metrics.mean_squared_error(y_test, y_pred)
partial_rmse = np.sqrt(partial_mse)

print(f"Partial MSE (Test Set): {partial_mse:.4f}")
print(f"Partial RMSE (Test Set): {partial_rmse:.4f}")
```

```python
# Calculate Full MSE and RMSE across the entire dataset
# Define the full dataset
X_full = sm.add_constant(study_hours)

# Predict using our trained model
y_full_pred = trained_model.predict(X_full)

# Calculate metrics
full_mse = metrics.mean_squared_error(Exam_score, y_full_pred)
full_rmse = np.sqrt(full_mse)

print(f"Full MSE (Entire Dataset): {full_mse:.4f}")
print(f"Full RMSE (Entire Dataset): {full_rmse:.4f}")
```

```python
# Define the correct 8-sample features matrix with constant
X_full_constant = sm.add_constant(study_hours)

# Calculate predictions using the full model (trained on all 8 samples)
y_full_model_pred = full_model.predict(X_full_constant)

# Calculate metrics using full_model predictions and original Exam_scores
full_model_mse = metrics.mean_squared_error(Exam_score, y_full_model_pred)
full_model_rmse = np.sqrt(full_model_mse)

print(f"Full Model MSE (8-sample trained model): {full_model_mse:.4f}")
print(f"Full Model RMSE (8-sample trained model): {full_model_rmse:.4f}")
```

```python
df=pd.read_csv('Multiplt_Linear_Regression.csv',sep=',')
df.head()
```

TRained Test  Linrear  regression models

```python
x=df[['Study_Hours','Attendance_Percent','Assignments_Completed','Previous_Score','Sleep_Hours']]
y=df['Exam_Score']
```

# **Classic model**

```python
# 1. Add constant to independent variables
x_const = sm.add_constant(x)

# 2. Fit the classic OLS model on the entire dataset
classic_model = sm.OLS(y, x_const).fit()
print(classic_model.summary())

# 3. Predict on the entire dataset
y_pred_full = classic_model.predict(x_const)

# 4. Calculate MSE and RMSE
classic_mse = metrics.mean_squared_error(y, y_pred_full)
classic_rmse = np.sqrt(classic_mse)

print("\n--- Classic Model Evaluation (Entire Dataset) ---")
print(f"MSE:  {classic_mse:.4f}")
print(f"RMSE: {classic_rmse:.4f}")
```

# **Train  and  Test  Model**

```python
# 1. Split the dataset into train and test sets (80% train, 20% test)
X_train, X_test, y_train, y_test = model_selection.train_test_split(x, y, test_size=0.2, random_state=42)

# 2. Add constant term to features (for statsmodels OLS)
X_train_const = sm.add_constant(X_train)
X_test_const = sm.add_constant(X_test)

# 3. Fit the model
mlr_model = sm.OLS(y_train, X_train_const).fit()
print(mlr_model.summary())

# 4. Predict on the test set
y_pred_test = mlr_model.predict(X_test_const)

# 5. Calculate MSE and RMSE
test_mse = metrics.mean_squared_error(y_test, y_pred_test)
test_rmse = np.sqrt(test_mse)

print("\n--- Evaluation Metrics on Test Set ---")
print(f"MSE:  {test_mse:.4f}")
print(f"RMSE: {test_rmse:.4f}")
```

#**Comparison Between Classic and Trained Model**

```python
# Create a comparison DataFrame
comparison_data = {
    'Metric': ['Mean Squared Error (MSE)', 'Root Mean Squared Error (RMSE)'],
    'Classic Model (Full Dataset)': [classic_mse, classic_rmse],
    'Train-Test Model (Test Set)': [test_mse, test_rmse]
}

comparison_df = pd.DataFrame(comparison_data)

# Display the table
print("Comparison Between Classic and Trained Model:")
print("==================================================")
print(comparison_df.to_string(index=False))
print("==================================================")
```
